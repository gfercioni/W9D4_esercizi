## Esercizio 1 · Costrutto if/elif/else (semplice · 20 minuti)

**Dati:** temperatura = 78 (in gradi) rilevata su un macchinario, con tre soglie: allarme se superiore a 80, attenzione se superiore a 60, normale altrimenti.

1. Scrivere un costrutto if/elif/else che stampi "Allarme", "Attenzione" o "Normale" in base al valore di temperatura, usando gli operatori di confronto corretti.
2. Modificare il valore di temperatura con altri due valori a scelta, uno che attivi il ramo elif e uno che attivi il ramo else, eseguendo di nuovo il codice per ciascuno.
3. Spiegare, in una frase, perché in nessun caso vengono eseguiti più di uno tra i rami if, elif ed else.

**Definizione richiesta:** scrivere la definizione di costrutto condizionale e indicare la differenza tra elif ed else.

**Criterio di completamento:** il codice è eseguito con tre valori diversi di temperatura e per ciascuno è stampato un solo messaggio tra i tre possibili.

Esercizio 1 punto 1

```python
temperatura = 78

if temperatura > 80:
    print("Allarme")
elif temperatura > 60:
    print("Attenzione")
else: 
    print("Normale")
```

Esercizio 1 punto 2

```python
temperatura = 81

if temperatura > 80:
    print("Allarme")
elif temperatura > 60:
    print("Attenzione")
else: 
    print("Normale")

temperatura = 61

if temperatura > 80:
    print("Allarme")
elif temperatura > 60:
    print("Attenzione")
else: 
    print("Normale")
```

Esercizio 1 · Punto 3
Le condizioni vengono valutate in ordine dall'alto verso il basso: appena una risulta vera viene eseguito il suo blocco e tutti i rami successivi vengono saltati, mentre l'else viene eseguito solo se nessuna condizione precedente è vera.

### Esercizio 1 · Definizione
**Costrutto condizionale:** struttura che permette di eseguire blocchi di codice diversi in base al valore di una condizione booleana (True o False). In Python si scrive con le parole chiave if, elif ed else, e viene eseguito al massimo un solo blocco tra quelli previsti.

**Differenza tra elif ed else:** elif ha una propria condizione, che viene valutata solo se tutte le condizioni precedenti sono false, ed esegue il suo blocco solo se questa condizione è vera. else invece non ha nessuna condizione: si trova sempre in fondo ed esegue il suo blocco solo quando tutte le condizioni precedenti sono risultate false.
