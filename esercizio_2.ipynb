## Esercizio 2 · Ciclo for (semplice · 20 minuti)

**Dati:** articoli = ["penna", "quaderno", "zaino"] e prezzi = {"penna": 1.5, "quaderno": 3.0, "zaino": 25.0}.

1. Usare un ciclo for per stampare ogni elemento della lista articoli, uno per riga.
2. Usare un ciclo for con il metodo .items() per stampare ogni coppia articolo-prezzo del dizionario prezzi.
3. All'interno del ciclo sul dizionario, usare un if per stampare un messaggio aggiuntivo solo per gli articoli con prezzo superiore a 10.

**Definizione richiesta:** scrivere la definizione di ciclo for e spiegare la differenza tra iterare direttamente su un dizionario e iterare con .items().

**Criterio di completamento:** sono stampati tutti gli articoli, tutte le coppie articolo-prezzo, e il messaggio aggiuntivo compare solo per gli articoli sopra soglia.

esercizio 2 punto 1

```python
articoli = ["penna", "quaderno", "zaino"]
for articolo in articoli:
    print(articolo)
```

## esercizio 2 punto2

```python
prezzi = {"penna": 1.5, "quaderno": 3.0, "zaino": 25.0}
for articolo, prezzo in prezzi.items():
    print(articolo, prezzo)
```

## esercizio 2 punto 3

```python
for articolo, prezzo in prezzi.items():
    if prezzo > 10:
        print(f"Articolo: {articolo}, Prezzo: {prezzo}")
    else:
        print("prezzo inferiore a 10")
```
