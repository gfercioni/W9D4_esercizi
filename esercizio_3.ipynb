esercizio 3 punto 3

```python
def total_price(price, quantity, sconto = 0):
    prezzo_senza_sconto = price * quantity
    if sconto != 0:
        prezzo_con_sconto = prezzo_senza_sconto - (prezzo_senza_sconto * sconto / 100)
        return prezzo_con_sconto   
    else:
       return prezzo_senza_sconto

print(total_price(quantity = 9, price = 7, sconto = 10))

print(total_price(3, 6, 30))
print(total_price(100, 6, 15))
```
