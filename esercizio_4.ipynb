## Esercizio 4 · Ciclo while (avanzato · 30 minuti)

**Dati:** un capitale iniziale di 1000 euro che cresce del 5% ogni anno; si vuole sapere dopo quanti anni supera 1500 euro.

1. Scrivere un ciclo while che aggiorni il capitale di anno in anno finché non supera 1500 euro, aggiornando a ogni iterazione sia il capitale sia il contatore degli anni.
2. Stampare il capitale e il numero di anni trascorsi a ogni iterazione del ciclo.
3. Indicare quale istruzione, se rimossa dal corpo del ciclo, trasformerebbe questo while in un loop infinito, e per quale motivo.

**Definizione richiesta:** scrivere la definizione di ciclo while e spiegare la differenza rispetto al ciclo for in termini di numero di iterazioni noto o non noto in anticipo.

**Criterio di completamento:** il ciclo termina da solo, il numero di anni necessari è riportato come valore numerico, e l'istruzione critica per evitare il loop infinito è indicata correttamente.

eserizio 4 punto 1 & 2

```python
capitale = 1000
anno_fiscale = 0


while capitale <= 1500:
    anno_fiscale += 1
    creascita_annuale = capitale/100*5
    capitale = capitale + creascita_annuale
    print(f"nell'anno fiscale {anno_fiscale} il capitale aziendale è {capitale}")
print(f"Il capitale aziendale supera i 1500€ dopo {anno_fiscale} anni")
```

```python
capitale = 1000
anno_fiscale = 2026


while capitale <= 1500:
    anno_fiscale += 1
    creascita_annuale = capitale/100*5
    capitale = capitale + creascita_annuale
    print(f"nell'anno fiscale {anno_fiscale} il capitale aziendale è {capitale}")
print(f"Il capitale aziendale supera i 1500€ dopo {anno_fiscale - 2026} anni")
```
