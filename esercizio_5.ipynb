## Esercizio 5 · List comprehension (avanzato · 30 minuti)

**Dati:** voti = [5, 8, 6, 9, 4, 7, 10].

1. Scrivere un ciclo for con append() che costruisca una nuova lista contenente solo i voti maggiori o uguali a 6.
2. Scrivere una list comprehension che ottenga lo stesso risultato del punto 1, con la sintassi [espressione for elemento in iterabile if condizione].
3. Scrivere una seconda list comprehension che moltiplichi per 10 ciascun voto della lista voti, senza alcuna condizione di filtro.

**Definizione richiesta:** scrivere la definizione di list comprehension ed elencare i casi in cui è preferibile tornare a un ciclo for.

**Criterio di completamento:** i risultati dei punti 1 e 2 sono stampati e coincidono elemento per elemento; il risultato del punto 3 ha la stessa lunghezza della lista voti.

## Esercizio 5 punto 1

```python
voti = [5, 8, 6, 9, 4, 7, 10]

voti_sufficienti = []

for number in voti:
    if number >= 6:
        voti_sufficienti.append(number)
print(voti_sufficienti)
```

## esercizio 5 punto 2

```python
## [espressione for elemento in iterabile if condizione]
voti_sufficienti = [voto for voto in voti if voto >= 6]
print(voti_sufficienti)
```

## esercizio 5 punto 3

```python
voti_moltiplicati = [voto * 10 for voto in voti]
print(voti_moltiplicati)

print(len(voti))
print(len(voti_moltiplicati))

if len(voti) == len(voti_moltiplicati):
    print("VITTORIA")
```
