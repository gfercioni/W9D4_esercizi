## Esercizio 6 punto 1

```python
import math

```

```python
numeri = [5,67,3,456,12]
```

```python
def radice_arrotondata(numero, decimali=2):
    radice_non_rotonda = math.sqrt(numero)  # sqrt della libreria math: radice quadrata
    return round(radice_non_rotonda, decimali)

print(radice_arrotondata(83, 2))
```

## Esercizio 6 punto 2

```python
lista_radici = []
for numero in numeri:
    valore = radice_arrotondata(numero)
    lista_radici.append(valore)
    print(valore)

print(lista_radici)
```
